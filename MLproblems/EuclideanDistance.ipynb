# Implement Euclidean Distance

**Easy** &nbsp;·&nbsp; TensorTonic &nbsp;·&nbsp; `Linear Algebra`

The Euclidean distance (L2 distance) between two equal-length vectors `x`, `y` is:

$$d(x, y) = \sqrt{\sum_{i} (x_i - y_i)^2}$$

---

**Example 1:**

```
Input:  x = [3, 4], y = [0, 0]
Output: 5.0
sqrt((3-0)^2 + (4-0)^2) = sqrt(9 + 16) = sqrt(25) = 5
```

**Example 2:**

```
Input:  x = [1, 2, 3], y = [4, 5, 6]
Output: 5.196152422706632
sqrt((1-4)^2 + (2-5)^2 + (3-6)^2) = sqrt(9 + 9 + 9) = sqrt(27) = 5.196
```

**Example 3:**

```
Input:  x = [0, 0], y = [0, 0]
Output: 0.0
```
