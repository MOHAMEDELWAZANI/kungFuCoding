# Implement Manhattan Distance

**Easy** &nbsp;·&nbsp; TensorTonic &nbsp;·&nbsp; `Linear Algebra`

The Manhattan distance (L1 distance) between two equal-length vectors `x`, `y` is:

$$d(x, y) = \sum_{i} |x_i - y_i|$$

It is called *city-block* distance because it measures horizontal travel plus
vertical travel along a grid — you cannot cut diagonally through the buildings.

---

**Example 1:**

```
Input:  x = [1, 2, 3], y = [2, 4, 6]
Output: 6.0
|1-2| + |2-4| + |3-6| = 1 + 2 + 3 = 6
```

**Example 2:**

```
Input:  x = [-1, -2], y = [1, 2]
Output: 6.0
|-1-1| + |-2-2| = 2 + 4 = 6
```

**Example 3:**

```
Input:  x = [0, 0], y = [0, 0]
Output: 0.0
```
